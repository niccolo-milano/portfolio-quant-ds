# Titanic Dataset — Initial Exploration

## Objective

The objective of this notebook is to explore the structure and quality of the Titanic dataset, identify missing values and answer initial questions using pandas selection and filtering.

```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
# Reading of Dataset
titanic = pd.read_csv("../data/train.csv")
```

## Dataset Overview

The following checks provide an initial overview of the dataset structure, dimensions and available variables.

```python
# Initial Diagnosis
display(titanic.head())

print("Number of rows and columns: ", titanic.shape)
print("Column names: ", titanic.columns)
```

The following checks show the data types associated with each variable in the dataset.

```python
# Display dataset information
titanic.info()

# Display data types
display(titanic.dtypes.to_frame(name="data_type"))
```

## Data Quality

This section examines missing values and duplicated rows to assess the initial quality of the dataset.

```python
# Computing of missing values and their percentage for every column on the total of rows
missing_summary = pd.DataFrame({"missing_values" : titanic.isnull().sum(), 
                                "missing_percentage" : titanic.isnull().mean() * 100})
missing_summary = (missing_summary.sort_values("missing_percentage", ascending = False))
missing_summary["missing_percentage"] = (missing_summary["missing_percentage"].round(2))
display(missing_summary)
# Computing the number of duplicated rows
print("The number of duplicated rows is: ", titanic.duplicated().sum())
```

## Initial Observations

`Cabin` has the highest proportion of missing values, with approximately 77% of its observations unavailable. Although this may suggest that the variable could be removed, its potential information content should be investigated before making a final decision.

`Age` is missing for approximately 20% of passengers and may require an appropriate imputation strategy in later stages. `Embarked` contains only two missing observations and is therefore a more limited data-quality issue.

The dataset contains no fully duplicated rows. However, missing values are concentrated in three variables.

## Filtering

This section creates five filtered subsets of the Titanic dataset to examine specific passenger groups and validate some of the initial observations.

```python
# 1. Passengers who survived
passengers_survived = titanic.loc[titanic["Survived"] == 1]
display(passengers_survived.head())
print(passengers_survived.shape)
```

```python
# 2. Female passengers who survived
female_survivors = titanic.loc[(titanic["Survived"] == 1) & (titanic["Sex"] == "female")]
display(female_survivors.head())
print(female_survivors.shape)
```

```python
# 3. First-class passengers who did not survive
first_class_not_survived = titanic.loc[(titanic["Survived"] == 0) & (titanic["Pclass"] == 1)]
display(first_class_not_survived.head())
print(first_class_not_survived.shape)
```

```python
# 4. Passengers younger than 18
minor_age_passengers = titanic.loc[titanic["Age"] < 18]
display(minor_age_passengers.head())
print(minor_age_passengers.shape)
```

```python
# 5. Passengers with missing age values
missing_age_passengers = titanic.loc[titanic["Age"].isnull()]
display(missing_age_passengers.head())
print(missing_age_passengers.shape)

print("Consistency check:", missing_age_passengers.shape[0] == titanic["Age"].isnull().sum())
```

## Observations

### Why might `PassengerId` be of limited analytical value?

Although `PassengerId` is stored as a numerical variable, it does not represent a measurable characteristic of the passenger. Its primary purpose is to uniquely identify each observation.

After verifying that the values are unique and contain no duplicates, the column could be used as the dataset index rather than as an analytical feature.

print("PassengerId is unique:", titanic["PassengerId"].is_unique)

print("Number of unique IDs:", titanic["PassengerId"].nunique())

We could set it up as the index of the dataset using:

titanic = titanic.set_index("PassengerId")

### Why is it risky to clean the data before analyzing it?

Cleaning data before understanding its structure may remove useful information or introduce unintended bias.

Missing values do not always represent a simple lack of information. In some cases, the fact that a value is missing may itself be associated with a passenger characteristic or with the process through which the data was collected.

For this reason, the dataset should be explored before deciding whether missing observations should be removed, imputed or transformed into new features.
